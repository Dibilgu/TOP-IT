# Лабораторная работа 3. Пользовательские функции

## Задание

Напишите функцию для вычисления суммы квадратов натуральных чисел от **a** до **b**, где **a** и **b** – натуральные числа. Продемоснстрируйте использование объявленной функции.

## Решение

### Объявление функции

```python
def lost(n):
  count = 0
  for i in range(0,n+1):
    if i % 2 == 0:
      count += i
  return count
```

### Демонстрация использования функции

```python
x = 50
print('Сумма четных чисел от 0 до', x,f': {lost(x)}')
x = 127
print('Сумма четных чисел от 0 до', x,f': {lost(x)}')
x = 391
print('Сумма четных чисел от 0 до', x,f': {lost(x)}')
```
