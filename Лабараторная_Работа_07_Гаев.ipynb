# Исходные данные

```python
file=open('p7_data_12.csv')
header=file.readline().strip().split(',')
all_employee=[]

for line in file:
    data=line.strip().split(',')
    employee={}
    for i, key in enumerate(header):
        employee[key]=data[i]
    all_employee.append(employee)
```

# Решение задачи №1
Определите сколько рабочих получают больше, чем вырабатывают
продукции.


```python
all_salary=[]
for line in all_employee:
    all_salary.append(float(line['Salary'][1:]))

up_salary=[]
mid_salary=sum(all_salary)/len(all_salary)

for salary in all_salary:
    if salary>mid_salary:
        up_salary.append(salary)
print('Количество сотрудников с зарплатой выше среднего:',len(up_salary),sep='')
```

# Решение задачи №2
Суммарный объем произведенной продукции (в валюте) без
категории.

```python
product=[]

for line in all_employee:
    count=int(line['Count'])*float(line['Price'][1:])
    product.append(count)

print('Объём произведенной продукции в валюте: $',sum(product),sep='')
```

# Решение задачи №3
Суммарный объем в валюте произведенной продукции по каждой из
4-х категорий.

```python
product_A,product_B,product_C,product_D=[],[],[],[]

for line in all_employee:
    count=int(line['Count'])*float(line['Price'][1:])
    category=line['Category']
    if category=='A':
        product_A.append(count)
    if category=='B':
        product_B.append(count)
    if category=='C':
        product_C.append(count)
    if category=='D':
        product_D.append(count)
print("Объём произведенной продукции А: $",sum(product_A),sep='')
print("Объём произведенной продукции B: $",sum(product_B),sep='')
print("Объём произведенной продукции C: $",sum(product_C),sep='')
print("Объём произведенной продукции D: $",sum(product_D),sep='')
```

# Решение задачи №4
Самый эффективный сотрудник: с максимальной разницей между
произведенным продуктом и зарплатой.


```python
best_employee=None
max_diff = float('-inf')

for employee in all_employee:
    product=int(employee['Count'])*float(employee['Price'][1:])
    salary=float(employee['Salary'][1:])
    diff=product-salary

    if diff>max_diff:
        max_diff=diff
        best_employee=employee['Name']

print('Максимальная разница: %.2f' % max_diff)
print('Самый эффективный сотрудник:', best_employee)
```

# Вопрос №4
Напишите код для поиска ключей словаря d, для которых значения
превышают 45:
p = {4:10, 12:100, 35:3, 34:56, 56:11, 45:’abc’}

```python
p = {4:10, 12:100, 35:3, 34:56, 56:11, 45:'abc'}

f=[]
for key,value in p.items():
    if isinstance(value,(int,float)) and value>45:
        f.append(key)

print(f)
```
